```python
# ============================================================
# STEP: CREATE CLEAN SKILL PROFILE FOR BUSINESS INTELLIGENCE
# ============================================================

# Keep only the columns we need
skill_profile = bi_skills[
    [
        "Element ID",
        "Element Name",
        "Scale Name",
        "Data Value",
        "N"
    ]
].copy()

# Separate Importance and Level
importance = skill_profile[
    skill_profile["Scale Name"] == "Importance"
][
    ["Element ID", "Element Name", "Data Value"]
].copy()

level = skill_profile[
    skill_profile["Scale Name"] == "Level"
][
    ["Element ID", "Element Name", "Data Value"]
].copy()

# Rename values
importance = importance.rename(
    columns={"Data Value": "Importance"}
)

level = level.rename(
    columns={"Data Value": "Required_Level"}
)

# Merge importance and required level
clean_skill_profile = importance.merge(
    level,
    on=["Element ID", "Element Name"],
    how="inner"
)

# Sort by importance
clean_skill_profile = clean_skill_profile.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

print("Clean skill profile:")
print(f"Number of skills: {len(clean_skill_profile)}")

display(clean_skill_profile)
```
